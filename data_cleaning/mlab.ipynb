```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
```

```python
mlab_path = "/Users/priyankaramachandran/Desktop/UCD Courses/CN Project/dataset/MLab Sacramento.csv"

mlab = pd.read_csv(mlab_path)
mlab.head()
```

```python
# Standardize column names
mlab.columns = mlab.columns.str.lower().str.strip()
mlab.columns
```

```python
# Keep useful columns only
keep_cols = [
    "id",
    "date",
    "test_uuid",
    "test_time",
    "throughput_mbps",
    "min_rtt_ms",
    "loss_rate",
    "client_city",
    "client_state",
    "postal_code",
    "client_latitude",
    "client_longitude",
    "isp",
    "asn"
]

mlab = mlab[keep_cols].copy()
mlab.head()
```

```python
# Convert data types
mlab["date"] = pd.to_datetime(mlab["date"], errors="coerce")
mlab["test_time"] = pd.to_datetime(mlab["test_time"], errors="coerce")

mlab["throughput_mbps"] = pd.to_numeric(mlab["throughput_mbps"], errors="coerce")
mlab["min_rtt_ms"] = pd.to_numeric(mlab["min_rtt_ms"], errors="coerce")
mlab["loss_rate"] = pd.to_numeric(mlab["loss_rate"], errors="coerce")

mlab["client_latitude"] = pd.to_numeric(mlab["client_latitude"], errors="coerce")
mlab["client_longitude"] = pd.to_numeric(mlab["client_longitude"], errors="coerce")
mlab["asn"] = pd.to_numeric(mlab["asn"], errors="coerce")

mlab["postal_code"] = mlab["postal_code"].astype(str).str.replace(".0", "", regex=False).str.zfill(5)
mlab["client_city"] = mlab["client_city"].astype(str).str.strip()
mlab["client_state"] = mlab["client_state"].astype(str).str.strip()
mlab["isp"] = mlab["isp"].astype(str).str.strip()

mlab.info()
```

```python
# Basic filtering
before_rows = len(mlab)

mlab = mlab.dropna(subset=[
    "test_time",
    "throughput_mbps",
    "min_rtt_ms",
    "postal_code",
    "isp"
])

mlab = mlab[
    (mlab["throughput_mbps"] > 1) &
    (mlab["min_rtt_ms"].between(1, 500))
].copy()

after_rows = len(mlab)

print("Rows before cleaning:", before_rows)
print("Rows after cleaning:", after_rows)
print("Rows removed:", before_rows - after_rows)
```

```python
# Remove duplicates
before_dupes = len(mlab)

mlab = mlab.drop_duplicates().copy()

print("Duplicate rows removed:", before_dupes - len(mlab))
print("Final rows:", len(mlab))
```
